### **Return Value**

```python
# Modular Calculator
num1 = float(input("First number: "))
num2 = float(input("Second number: "))

# Functions, solves depending on operation
def add_numbers(num1, num2):
  sum = num1 + num2
  return sum

def subtract_numbers(num1, num2):
  difference = num1 - num2
  return difference

def multiply_numbers(num1, num2):
  product = num1 * num2
  return product

def divide_numbers(num1, num2):
  if num2 == 0:
    print("Cannot divide by zero. Try again.")
  else:
    quotient = num1/num2
    return quotient

# Asks user for operation
print("1-Addition, 2-Subtraction, 3-Multiplication, 4-Division")
operation = float(input("Operation(1, 2, 3, 4): "))

# Calculates based on operation and displays answer
if operation == 1:
  add_numbers(num1, num2)
  print(add_numbers(num1,num2))

if operation == 2:
  subtract_numbers(num1, num2)
  print(subtract_numbers(num1,num2))

if operation == 3:
  multiply_numbers(num1, num2)
  print(multiply_numbers(num1,num2))

if operation == 4:
  divide_numbers(num1, num2)
  print(divide_numbers(num1,num2))
```
